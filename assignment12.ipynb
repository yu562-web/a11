**Create and visualize your model below...**

````r
library(tidymodels)

rec <- recipe(mpg ~ displacement + cylinders, data = Auto) |>
  step_ns(displacement, deg_free = 3) |>
  step_mutate(cyl_cat = factor(cylinders)) |>
  step_dummy(cyl_cat) |>
  step_rm(cylinders) |>
  prep()

mod_fit <- linear_reg() |>
  set_engine("lm") |>
  fit(mpg ~ ., data = juice(rec))

mod_fit

predict(mod_fit, juice(rec)) |>
  bind_cols(Auto) |>
  ggplot(aes(x = displacement, y = mpg)) +
  geom_point(size = 1, alpha = 0.35) +
  geom_line(aes(y = .pred, color = factor(cylinders)),
            linewidth = 1) +
  labs(title = "Spline model of fuel economy",
       x = "Displacement (cubic inches)",
       y = "Fuel economy (mpg)",
       color = "Cylinders") +
  theme_minimal()
````
